```python

# Preserve the original dataset
adult = train.copy()

# Clean whitespace in text columns
cat_cols = adult.select_dtypes(
    include=["object", "string"]
).columns

for c in cat_cols:
    adult[c] = adult[c].astype("string").str.strip()

# Replace missing categorical values
for c in ["workclass", "occupation", "native_country"]:
    adult[c] = adult[c].fillna("Unknown")

# Create the first derived feature
adult["income_over_50k"] = adult["income"].eq(">50K")

# Validate the cleaning
print("Remaining missing values:")
print(adult[["workclass", "occupation", "native_country"]].isna().sum())

print("Rows before cleaning:", len(train))
print("Rows after cleaning:", len(adult))
print("Duplicate rows:", adult.duplicated().sum())
```

```python

adult["work_pattern"] = adult["hours_per_week"].apply(
    lambda h: "Part-time" if h < 35
    else "Full-time-like" if h <= 40
    else "Over 40 hours"
)

display(adult[["hours_per_week", "work_pattern"]].head(10))
```

```python

# Inspect duplicate records
duplicate_rows = adult[adult.duplicated(keep=False)]

print("Rows involved in duplicates:", len(duplicate_rows))
display(duplicate_rows.head(10))

# Compare original and cleaned duplicate counts
print("Original duplicate rows:", train.duplicated().sum())
print("Current duplicate rows:", adult.duplicated().sum())
```

```python

work_summary = (
    adult.groupby("work_pattern", dropna=False)
    .agg(
        records=("income_over_50k", "size"),
        over_50k_rate=("income_over_50k", "mean"),
        avg_hours=("hours_per_week", "mean"),
        median_age=("age", "median")
    )
    .reset_index()
)

work_summary["over_50k_pct"] = (
    work_summary["over_50k_rate"] * 100
).round(2)

work_summary = work_summary.sort_values(
    "over_50k_pct", ascending=False
)

display(work_summary)
```

```python

kpi = (
    adult.groupby(["workclass", "education"], dropna=False)
    .agg(
        records=("income_over_50k", "size"),
        over_50k_rate=("income_over_50k", "mean"),
        avg_hours=("hours_per_week", "mean"),
        median_age=("age", "median")
    )
    .reset_index()
)

kpi["over_50k_pct"] = (
    kpi["over_50k_rate"] * 100
).round(2)

top_20 = (
    kpi.query("records >= 30")
    .sort_values("over_50k_pct", ascending=False)
    .head(20)
)

display(top_20)
```

```python

# Confirm that the schemas match
assert train.columns.equals(test.columns)

# Stack the two partitions and preserve their source
combined = pd.concat(
    [
        train.assign(source="train"),
        test.assign(source="test")
    ],
    ignore_index=True
)

print("Train rows:", len(train))
print("Test rows:", len(test))
print("Combined rows:", len(combined))

print("\nRows by source:")
print(combined["source"].value_counts())
```
