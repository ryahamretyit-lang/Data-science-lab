```python
job_summary = pd.crosstab(
    b["job"],
    b["subscribed"]
)

print(job_summary)
```

```python
job_summary["total"] = job_summary.sum(axis=1)

job_summary["subscription_rate_pct"] = (
    job_summary[True] / job_summary["total"] * 100
).round(2)

print(job_summary)
```

```python
top_job = job_summary["subscription_rate_pct"].idxmax()
top_rate = job_summary["subscription_rate_pct"].max()

print("Job category with the highest subscription rate:", top_job)
print("Subscription rate:", top_rate, "%")
```

```python
check_rates = (
    job_summary[True] / job_summary["total"] * 100
).round(2)

print("All subscription rates match:",
      check_rates.equals(job_summary["subscription_rate_pct"]))
```

```python
job_summary_sorted = job_summary.sort_values(
    by="subscription_rate_pct",
    ascending=False
)

print(job_summary_sorted)
```

```python
rates = job_summary_sorted["subscription_rate_pct"]

print("Rates sorted from highest to lowest:",
      rates.is_monotonic_decreasing)
```

```python
job_summary_final = job_summary_sorted[
    ["total", True, "subscription_rate_pct"]
].rename(columns={
    True: "subscribed_count",
    "subscription_rate_pct": "subscription_rate_pct"
})

print(job_summary_final)
```

```python
resp = (
    b.groupby(["job", "contact"], dropna=False)
     .agg(
         contacts=("subscribed", "size"),
         successes=("subscribed", "sum"),
         response_rate=("subscribed", "mean"),
         avg_balance=("balance", "mean")
     )
     .reset_index()
)

resp["response_pct"] = resp["response_rate"] * 100

print(resp.head(10))
```

```python
resp.query("contacts >= 50").sort_values(
    "response_pct",
    ascending=False
)
```

```python
resp_filtered = resp.query("contacts >= 50").sort_values(
    "response_pct",
    ascending=False
)

assert resp_filtered["contacts"].ge(50).all()
assert resp_filtered["response_pct"].is_monotonic_decreasing

print("Both validation checks passed.")
```

### Cleaning Decision Log — UCI Bank Marketing

| Cleaning or transformation | Decision and reason |
|---|---|
| Preserve original data | Kept `bank` unchanged and used `b` as the working copy. |
| Standardize text | Removed surrounding whitespace and converted text categories to lowercase for consistency. |
| Create `subscribed` | Converted `y` into a Boolean feature: `True` for yes and `False` for no. |
| Create `has_previous_contact` | Used `previous > 0` to identify observations with previous campaign contacts. |
| Create `balance_band` | Used quartile-based bins to group account balances for analysis. |
| Missing values | Found zero actual missing cells in the original dataset; did not apply unnecessary blanket imputation. |
| Duplicate rows | Found zero duplicate rows in the cleaned working dataset. |
| Preserve `unknown` | Retained `unknown` as a recorded category rather than treating it as a pandas missing value. |

**Limitation:** Standardizing text improves consistency but does not verify that every recorded category is factually correct. The `subscribed` outcome records the campaign result; it does not establish that campaign contact caused the outcome.

## Management Note — UCI Bank Marketing Analysis

### Objective and data preparation

This analysis examined the UCI Bank Marketing dataset to summarize term-deposit subscription outcomes across customer job categories and campaign contact methods. The original dataset contained 45,211 contact observations and 17 columns. Initial checks found no actual missing cells and no fully duplicated rows. The original `bank` DataFrame was preserved, while a separate working copy, `b`, was used for text standardization and feature engineering.

### Key findings

The job-level summary showed that students had the highest observed subscription response rate, approximately 28.68%, followed by retired customers at 22.79%. Blue-collar workers had the lowest observed rate, approximately 7.27%. These results identify differences worth further investigation, but they do not establish that job category causes subscription outcomes. Response rates also varied across combinations of job category and contact method.

### Filtering and assumptions

A focused campaign segment contained 6,902 observations, selected using cellular contact, at least one previous contact, and exclusion of student, retired, and unknown job categories. This represented approximately 15.3% of the original dataset. Findings from this subset should not be generalized automatically to the full population.

### Validation and limitations

Derived features were checked against their source columns. Grouped summaries were validated against overall observation and subscriber counts. The job-family lookup was merged using a many-to-one validation, and exported CSV files were checked after reading them back. The data describes campaign contact observations, not necessarily unique customers. Further analysis should consider confounding variables, campaign duration, and selection effects before making operational or causal claims.
